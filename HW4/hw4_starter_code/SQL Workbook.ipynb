## Problem 1 SQL Queries

### Part (a) - All male Olympic athletes from Team Canada sorted alphabetically

```python
q = "SELECT * " \
    "FROM athlete " \
    "WHERE Team = 'Canada' AND Sex = 'M' " \
    "ORDER BY Name;"

pd.read_sql(q, conn)
```

### Part (b) - Top five Olympic games that awarded the most medals

```python
q = "SELECT g.Games, g.Year, g.Season, g.City, COUNT(*) AS NumMedals " \
    "FROM games g " \
    "JOIN medals m ON g.id = m.games_id " \
    "GROUP BY g.id, g.Games, g.Year, g.Season, g.City " \
    "ORDER BY NumMedals DESC " \
    "LIMIT 5;"

pd.read_sql(q, conn)
```

### Part (c) - Top 3 NOC regions that won the most gold medals

```python
q = "SELECT n.region, COUNT(*) AS NumGoldMedals " \
    "FROM noc n " \
    "JOIN athlete a ON n.NOC = a.NOC " \
    "JOIN medals m ON a.id = m.athlete_id " \
    "WHERE m.Medal = 'Gold' " \
    "GROUP BY n.region " \
    "ORDER BY NumGoldMedals DESC " \
    "LIMIT 3;"

pd.read_sql(q, conn)
```

### Part (d) - Top five athletes who won the most medals in the same Sport

```python
q = "SELECT a.Name, e.Sport, COUNT(*) AS NumMedals " \
    "FROM athlete a " \
    "JOIN medals m ON a.id = m.athlete_id " \
    "JOIN event e ON m.event_id = e.id " \
    "GROUP BY a.id, a.Name, e.Sport " \
    "ORDER BY NumMedals DESC " \
    "LIMIT 5;"

pd.read_sql(q, conn)
```

### Part (e) - Athletes who won all three medals (bronze, silver, gold) in the same event

```python
q = "SELECT a.Name, e.Event " \
    "FROM athlete a " \
    "JOIN medals m ON a.id = m.athlete_id " \
    "JOIN event e ON m.event_id = e.id " \
    "GROUP BY a.id, a.Name, e.id, e.Event " \
    "HAVING COUNT(DISTINCT m.Medal) = 3 " \
    "   AND SUM(CASE WHEN m.Medal = 'Gold' THEN 1 ELSE 0 END) > 0 " \
    "   AND SUM(CASE WHEN m.Medal = 'Silver' THEN 1 ELSE 0 END) > 0 " \
    "   AND SUM(CASE WHEN m.Medal = 'Bronze' THEN 1 ELSE 0 END) > 0 " \
    "ORDER BY a.Name;"

pd.read_sql(q, conn)
```
